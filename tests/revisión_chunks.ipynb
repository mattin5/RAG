```python
import json
cs = [json.loads(l) for l in open("../data/processed/chunks_min200.jsonl", encoding="utf-8")]
cortos = sorted(cs, key=lambda c: c["n_tokens"])[:10]
for c in cortos:
    print(c["n_tokens"], "|", c["breadcrumb"], "|", repr(c["text"][:80]))
```

```python
import numpy as np
t = np.array([c["n_tokens"] for c in cs])
print(np.percentile(t, [5, 25, 50, 75, 90, 95]))
print("por debajo de 100:", (t < 100).sum())
```
