```python
import json
from pathlib import Path

chunks_path = Path("../data/processed/chunks.jsonl")

# Cambia este valor por el chunk_id que quieras buscar
target_chunk_ids = ["mlflow/docs/docs/classic-ml/traditional-ml/tutorials/hyperparameter-tuning/notebooks/index#0000-hyperparameter-tuning-with-mlflow-and-child-runs---notebooks",
                    "mlflow/docs/docs/genai/getting-started/try-assistant#0000-mlflow-ai-assistant"]

def get_chunks_by_id(path, chunk_id):
    matches = []
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            chunk = json.loads(line)
            if chunk.get("chunk_id") == chunk_id:
                matches.append(chunk)
    return matches

for chunk_id in target_chunk_ids:
    chunk = get_chunks_by_id(chunks_path, chunk_id)

    
    print(json.dumps(chunk, ensure_ascii=False))
    print(chunk[0]["text"])
```
